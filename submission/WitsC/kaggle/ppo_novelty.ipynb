```python
import base64, hashlib, io, zipfile
from pathlib import Path
payload = base64.b64decode('')
assert hashlib.sha256(payload).hexdigest() == '5acafa6208e859e2de4d66a7a0296b0ac1c362a9d17b1143be87da4d63dbfefa'
root = Path('/tmp/witsc_assets')
root.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(io.BytesIO(payload)) as archive:
    assert set(archive.namelist()) == {'weights.npz', 'model.json', 'my_agent.py', 'runtime.py'}
    archive.extractall(root)
del payload
```
